```python
new_datasets=pd.read_excel("newest_wur_data_uuid_4tu.xlsx")
old_datasets=pd.read_excel("old_datasets_less_2023_4tu_uuid.xlsx")
ref_dataset=pd.read_excel("all_data_wur_uuid_4tu.xlsx")
```

```python

old_datasets.columns
```

```python


merged_dataset_new_old=pd.merge(old_datasets, new_datasets, on="Title Dataset", how="outer", indicator=False)
merged_dataset_ref=pd.merge(ref_dataset, merged_dataset_new_old, on="Title Dataset", how="outer", indicator=False)

cols = ["DOI Dataset_x","doi","DOI Dataset_y"]
merged_dataset_ref["DOI Dataset"] = merged_dataset_ref[cols].bfill(axis=1).iloc[:, 0]

## Remove the original columns
merged_dataset_ref.drop(columns=cols, inplace=True)

## Replace container string with nothing

merged_dataset_ref["container"] = merged_dataset_ref["container"].str.replace("container:", "", regex=False)

cols_1=["container","UUID 4TU_y","UUID 4TU_x"]
merged_dataset_ref["UUID 4TU"] = merged_dataset_ref[cols_1].bfill(axis=1).iloc[:, 0]

merged_dataset_ref.drop(columns=cols_1, inplace=True)

## Merge UUID Research output column
cols_2=["UUID Research output_x","UUID Research output_y"]
merged_dataset_ref["UUID Research output"] = merged_dataset_ref[cols_2].bfill(axis=1).iloc[:, 0]
merged_dataset_ref.drop(columns=cols_2, inplace=True)

## Merge Type Research output column
cols_3=["Type Research output_x","Type Research output_y"]
merged_dataset_ref["Type Research output"] = merged_dataset_ref[cols_3].bfill(axis=1).iloc[:, 0]
merged_dataset_ref.drop(columns=cols_3, inplace=True)  

## Merge UUID Dataset column
cols_4=["UUID Dataset_x","UUID Dataset_y"]
merged_dataset_ref["UUID Dataset"] = merged_dataset_ref[cols_4].bfill(axis=1).iloc[:, 0]
merged_dataset_ref.drop(columns=cols_4, inplace=True)

## Merge DOIs (Digital Object Identifiers) Research output column
cols_5=["DOIs (Digital Object Identifiers) Research output_x","DOIs (Digital Object Identifiers) Research output_y"]
merged_dataset_ref["DOI Research output"] = merged_dataset_ref[cols_5].bfill(axis=1).iloc[:, 0]
merged_dataset_ref.drop(columns=cols_5, inplace=True)

## Merge Title Research output column
cols_6=["Title Research output_x","Title Research output_y"]
merged_dataset_ref["Title Research output"] = merged_dataset_ref[cols_6].bfill(axis=1).iloc[:, 0]
merged_dataset_ref.drop(columns=cols_6, inplace=True)

## Merge Date made available column
cols_7=["Date made available_x","Date made available_y"]
merged_dataset_ref["Date made available"] = merged_dataset_ref[cols_7].bfill(axis=1).iloc[:, 0]
merged_dataset_ref.drop(columns=cols_7, inplace=True)
```

```python
merged_dataset_ref.to_excel("merged_dataset_ref_with_doi.xlsx", index=False)
```
